```python
#import libraries
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
#load dataset
df = pd.read_csv('/content/messy_ecommerce_orders_10000_rows.csv')

print(df.head())
```

```python
#checking dataset
print("Dataset Shape:", df.shape)

print("\nColumn Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
#check missing value
df.isnull().sum()
```

```python
#convert qty into numeric
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
#filling missing values for numerical column
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['UnitPrice']= df['UnitPrice'].fillna(df['UnitPrice'].median())
print(df[['Quantity','UnitPrice']].isnull().sum())
```

```python
#for text columns
df['CustomerID'] = df['CustomerID'].fillna('unknown')
df['City'] = df['City'].fillna('Unknown')
df['PaymentMethod'] = df['PaymentMethod'].fillna('Unknown')
print(df[['CustomerID','City','PaymentMethod']].isnull().sum())
```

```python
#for missing dates
df['OrderDate'] = pd.to_datetime(df['OrderDate'],errors ='coerce',dayfirst = True)
print(df['OrderDate'].isnull().sum())
```

```python
#checking again missing values
print(df.isnull().sum())
```

```python
#check duplicate records
print("Duplicate rows:",df.duplicated().sum())
```

```python
#Remove duplicates
df = df.drop_duplicates()
print("duplicates after cleaning:", df.duplicated().sum())
```

```python
#correcting data types
print(df.dtypes)
```

```python
#converting quantity,unitprice,sales,& profit into numeric
numeric_cols = ['Quantity', 'UnitPrice','Sales', 'Profit']
for col in numeric_cols:
  df[col] = pd.to_numeric(df[col], errors='coerce')

#convert orderdate into date format
df['OrderDate'] = pd.to_datetime(df['OrderDate'], errors ='coerce',dayfirst = True)
print(df.dtypes)
```

```python
#finding invalid values
print("Negative Quantities:")
print((df['Quantity'] < 0).sum())

print("Negative Prices:")
print((df['UnitPrice'] < 0).sum())

print("Highest Unit price:")
print((df['UnitPrice'].max()))
```

```python
#replace negative qty & price with missing value
df.loc[df['Quantity'] <=0,'Quantity'] = np.nan
df.loc[df['UnitPrice'] <=0,'UnitPrice'] = np.nan

df['Quantity'] = df['Quantity'].fillna(
    df['Quantity'].median())

df['UnitPrice'] = df['UnitPrice'].fillna(
    df.groupby('Product')['UnitPrice'].transform('median'))

print(df[['Quantity', 'UnitPrice']].head(10))
print(df[['Quantity', 'UnitPrice']].isnull().sum())
```

```python
#handle outliers
def remove_outliers(group):
    Q1 = group['UnitPrice'].quantile(0.25)
    Q3 = group['UnitPrice'].quantile(0.75)

    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    group['UnitPrice'] = group['UnitPrice'].clip(lower=lower, upper=upper)
    return group
df = df.groupby('Product', group_keys=False).apply(remove_outliers)

#recalculate sales & profit
df['Sales'] = df['Quantity'] * df['UnitPrice']
df['Profit'] = df['Sales'] - (df['Quantity'] * df['UnitCost'])

print(df[['Quantity', 'UnitPrice', 'Sales', 'Profit']].describe())
```

```python
#remove unwanted text spaces
text_cols = ['Product', 'Category', 'City']
for col in text_cols:
  df[col] = df[col].str.strip().str.title()
df['PaymentMethod'] = df['PaymentMethod'].str.strip().str.upper()
print("Text standardization completed successfully!")
print(df[['Product', 'Category', 'City', 'PaymentMethod']].head())
```

```python
#handle missing dates
df = df.dropna(subset = ['OrderDate'])

#extract month & year(feature engineering)
df['Month'] = df['OrderDate'].dt.month
df['Year'] = df['OrderDate'].dt.year

#calculating profit margin
df['ProfitMargin'] = (df['Profit'] / df['Sales']) * 100

print(df.head())
print(df.info())
```

```python
#verifying
print("Final Dataset shape:",df.shape)

print("\nMissing Values:",df.isnull().sum())

print("\nDuplicate Rows:",df.duplicated().sum())

print("\nData Types:",df.dtypes)
```

```python
#export cleaned dataset
df.to_csv('clean_dataset.csv', index = False)
print("Clean dataset exported successfully")
```
